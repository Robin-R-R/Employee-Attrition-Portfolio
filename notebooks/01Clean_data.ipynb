```python
# Setup, load dataset and inspect shape

import pandas as pd
import matplotlib.pyplot as plt
df_row = pd.read_excel('~/Desktop/Original Dataset of Employee Attrition.xlsx')
df = df_row.copy()
df.shape
```

```python
# Check duplicated columns
# It may be two distinct employees with identical answers

print(df.duplicated().sum())
print(df.drop(columns=['ID']).duplicated().sum())
```

```python
# Standardize raw column names

df.columns = df.columns.str.strip()

df = df.rename(columns={
    'ID': 'id',
    'Attrition': 'attrition_raw',
    'Gender': 'gender_raw',
    'Age': 'age_raw',
    'Maritalstatus': 'marital_status_raw',
    'Academic_degree': 'academic_degree_raw',
    'Years_Experience': 'years_experience_raw',
    'Years_experience_lastorganization': 'years_last_org_raw',
    'Sector': 'sector_raw',
    'Department': 'department_raw',
    'JobTitle': 'job_title_raw',
    'MonthlySalary': 'monthly_salary_raw',
    'Allowances': 'allowances',
    'MedicalInsurance': 'medical_insurance_raw',
    'Bonus': 'bonus_raw',
    'OverTime': 'overtime_raw',
    'Payment_Overtime': 'payment_overtime_raw',
    'Rewards&Wages_Satisfaction': 'rewards_wages_satisfaction_raw',
    'Get_ Deserved_Promotion': 'deserved_promotion_raw',
    'Training_programs_ During_last_three_years': 'training_programs_raw',
    'Useful_Training_Programs': 'useful_training_raw',
    'Business_Travel': 'business_travel_raw',
    'Job_Support': 'job_support_raw',
    'Recognition': 'recognition_raw',
    'Emotional_Commitment': 'emotional_commitment_raw',
    'Job_Engagement': 'job_engagement_raw',
    'Distance_to_work': 'distance_to_work_raw',
    'Work_Live_Balance': 'work_life_balance_raw',
    'Physical_Stress': 'physical_stress_raw',
    'Psychological_Exhaustion': 'psychological_exhaustion_raw',
    'Job_Stability': 'job_stability_raw',
    'Health_Issues': 'health_issues_raw',
    'Environment_Satisfaction': 'environment_satisfaction_raw',
    'Job_Satisfaction': 'job_satisfaction_raw',
    'Job_Opportunities': 'job_opportunities_raw'
})
```

```python
# Inspect categorical value distributions

for col in df.select_dtypes(include='object').columns:
    print(f'\n {df[col].value_counts()}')
```

```python
# Text normalization
# Trim, lowercase, and fix value variants

import re
def normalize_text(x):
    if pd.isna(x):
        return x
    x = str(x).strip()
    x = re.sub(r'\s+', ' ', x)
    x = x.lower()
    x = x.replace('s.r.', 'sar').replace('s.r', 'sar')
    x = x.replace('manger', 'manager')
    return x

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].apply(normalize_text)
```

```python
# Export data quality audit

def audit_dataframe(df):
    rows = []
    for col in df.columns:
        s = df[col]
        rows.append({
            'column': col,
            'dtype': str(s.dtype),
            'n_missing': s.isna().sum(),
            'missing_rate': round(s.isna().mean(), 4),
            'n_unique': s.nunique(dropna=True),
            'sample_values': ' | '.join(map(str, s.dropna().unique()[:5]))
        })
    return pd.DataFrame(rows)

audit_df = audit_dataframe(df)
audit_df.to_csv('~/Desktop/Employee-Attrition- Portfolio/data/data_audit.csv', index=False)
```

```python
# Export value report

records = []
for col in df.select_dtypes(include='object').columns:
    vc = df[col].value_counts(dropna=False)
    for val, cnt in vc.items():
        records.append({
            'column': col,
            'value': val,
            'count': cnt
        })

value_df = pd.DataFrame(records)
value_df.to_csv('~/Desktop/Employee-Attrition- Portfolio/data/value_report.csv', index=False)
```

```python
# Create encoding rules

mappings = []

def add_mapping(column, mapping_dict, mtype):
    for raw, enc in mapping_dict.items():
        mappings.append({
            'column': column,
            'raw_value': raw,
            'encoded_value': enc,
            'type': mtype
        })

add_mapping('attrition_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('gender_raw', {'female': 0, 'male': 1}, 'binary')
add_mapping('medical_insurance_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('bonus_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('overtime_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('rewards_wages_satisfaction_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('deserved_promotion_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('useful_training_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('recognition_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('job_stability_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('health_issues_raw', {'no': 0, 'yes': 1}, 'binary')
add_mapping('job_opportunities_raw', {'no': 0, 'yes': 1}, 'binary')

add_mapping('job_support_raw', {'low': 1, 'medium': 2, 'high': 3}, 'ordinal')
add_mapping('emotional_commitment_raw', {'low': 1, 'medium': 2, 'high': 3}, 'ordinal')
add_mapping('environment_satisfaction_raw', {'low': 1, 'medium': 2, 'high': 3}, 'ordinal')
add_mapping('work_life_balance_raw', {'difficult': 1, 'medium': 2, 'easy': 3}, 'ordinal')
add_mapping('job_engagement_raw', {'difficult': 1, 'medium': 2, 'easy': 3}, 'ordinal')
add_mapping('distance_to_work_raw', {'close': 1, 'medium': 2, 'far': 3}, 'ordinal')
add_mapping('job_satisfaction_raw', {'not satisfied': 1, 'satisfied': 2, 'very satisfied': 3}, 'ordinal')
add_mapping('physical_stress_raw', {'no': 0, 'sometimes': 1, 'yes': 2}, 'ordinal')
add_mapping('psychological_exhaustion_raw', {'no': 0, 'sometimes': 1, 'yes': 2}, 'ordinal')

add_mapping('business_travel_raw', {
    'i do not travel for work': 0,
    'travel rarely': 1,
    'travel frequently': 2
}, 'ordinal')

add_mapping('training_programs_raw', {
    'i did not receive any training': 0,
    'from 1 to 3 training programs': 1,
    'from 4 to 6 training programs': 2,
    'from 7 training programs to more': 3
}, 'ordinal')

add_mapping('age_raw', {
    '21 to 30': 1,
    '31 to 40': 2,
    '41 to 50': 3,
    '51 to 60': 4
}, 'interval')

exp_map = {
    'less than 5 years': 1,
    'from 5 to 10 years': 2,
    'from 11 to 15 years': 3,
    'from 16 to 20 years': 4,
    'from 21 to 25 years': 5,
    'from 26 to 30 years': 6,
    'from 31 to 35 years': 7
}
add_mapping('years_experience_raw', exp_map, 'interval')
add_mapping('years_last_org_raw', exp_map, 'interval')

add_mapping('monthly_salary_raw', {
    'less than 5000 sar': 1,
    'from 5000 to 10000 sar': 2,
    'from 11000 to 15000 sar': 3,
    'from 16000 to 20000 sar': 4,
    'from 21000 to 25000 sar': 5,
    'from 26000 to 30000 sar': 6,
    'sar 31000 - and more': 7
}, 'interval')

add_mapping('academic_degree_raw', {
    'diploma or secondary': 1,
    "bachelor's": 2,
    "master's": 3,
    'ph.d': 4
}, 'ordinal')

mapping_df = pd.DataFrame(mappings)
mapping_df.to_csv('~/Desktop/Employee-Attrition- Portfolio/data/mapping.csv', index=False)
```

```python
# Apply mapping rules

for col in mapping_df['column'].unique():
    mp = mapping_df[mapping_df['column'] == col].set_index('raw_value')['encoded_value'].to_dict()
    new_col = col.replace('_raw', '')
    df[new_col] = df[col].map(mp)
    n_nan = df[new_col].isna().sum()
    if n_nan > 0:
        print(f"warning: {new_col} has {n_nan} values not mapping")
df.columns
```

```python
# Create midpoints for interval variables

age_mid_map = {
    '21 to 30': 25.5,
    '31 to 40': 35.5,
    '41 to 50': 45.5,
    '51 to 60': 55.5
}
df['age_mid'] = df['age_raw'].map(age_mid_map)

exp_mid_map = {
    'less than 5 years': 2.5,
    'from 5 to 10 years': 7.5,
    'from 11 to 15 years': 13,
    'from 16 to 20 years': 18,
    'from 21 to 25 years': 23,
    'from 26 to 30 years': 28,
    'from 31 to 35 years': 33
}
df['years_experience_mid'] = df['years_experience_raw'].map(exp_mid_map)
df['years_last_org_mid'] = df['years_last_org_raw'].map(exp_mid_map)

salary_mid_map = {
    'less than 5000 sar': 2500,  
    'from 5000 to 10000 sar': 7500,
    'from 11000 to 15000 sar': 13000,
    'from 16000 to 20000 sar': 18000,
    'from 21000 to 25000 sar': 23000,
    'from 26000 to 30000 sar': 28000,
    'sar 31000 - and more': 35000     
}
df['monthly_salary_mid'] = df['monthly_salary_raw'].map(salary_mid_map)
```

```python
# Check missing values of midpoint mappings

for col in ['age_mid', 'years_experience_mid', 'years_last_org_mid', 'monthly_salary_mid']:
    n_nan = df[col].isna().sum()
    print(f"{col}: {n_nan}  NaN")
```

```python
# Overtime status: 1 = no overtime, 2 = paid overtime, 3 = unpaid overtime, 4 = others

payment_map = {"yes": 1, "no": 0, "i don't have overtime": None}
df['payment_overtime'] = df['payment_overtime_raw'].map(payment_map)

df['ot_status'] = 1
df.loc[(df['overtime'] == 1) & (df['payment_overtime'] == 1), 'ot_status'] = 2
df.loc[(df['overtime'] == 1) & (df['payment_overtime'] == 0), 'ot_status'] = 3
df.loc[(df['overtime'] == 1) & (df['payment_overtime'].isna()), 'ot_status'] = 4

df['flag_overtime_payment_inconsistent'] = (
    (df['overtime'] == 1) & (df['payment_overtime'].isna())
).astype(int)
df['unpaid_overtime'] = (df['ot_status'] == 3).astype(int)

print(df['ot_status'].value_counts().sort_index())
print(pd.crosstab(df['ot_status'], df['overtime_raw']))

df = df.drop(columns=['payment_overtime'], errors='ignore')
```

```python
# # Merge rare sector category

sector_counts = df['sector_raw'].value_counts()
main_sectors = sector_counts[sector_counts >= 25].index.tolist()
df['sector_group'] = df['sector_raw'].apply(
    lambda x: x if x in main_sectors else 'other'
)
print(f"Sector groups: {df['sector_group'].nunique()}")
print(df['sector_group'].value_counts())
```

```python
# Merge rare department category

dept_counts = df['department_raw'].value_counts()
main_depts = dept_counts[dept_counts >= 25].index.tolist()
df['department_group'] = df['department_raw'].apply(
    lambda x: x if x in main_depts else 'other'
)
print(f"Department groups: {df['department_group'].nunique()}")
print(df['department_group'].value_counts())
```

```python
# Define direct title from ISCO document

DIRECT_MAP = {
    'teacher': 23, 'administrative manager': 12,
    'customer services representative': 42, 'nurse': 22,
    'receptionist': 42, 'accountant': 24,
    'sales representative': 52, 'salesman': 52, 'doctor': 22,
    'it specialist': 25, 'pharmacist': 22, 'hr specialist': 24,
    'marketing specialist': 24, 'hiring specialist': 24,
    'financial manager': 12, 'data entry operator': 41,
    'electrical engineer': 21, 'financial analyst': 24,
    'dentist': 22, 'lecturer': 23, 'lawyer': 26,
    'civil engineer': 21, 'secretariat': 41,
    'admin accountant': 24, 'engineer': 21, 'driver': 83,
    'mechanical engineer': 21, 'radiographer': 32,
    'nutrition specialist': 22, 'graphic designer': 21,
    'chemical engineer': 21, 'project engineer': 21,
    'social worker': 26,
    'manager assistant': 33, 'security and safety specialist': 54,
    'relationships specialist': 24, 'purchase specialist': 33,
    'relations manager': 12, 'production supervisor': 31,
    'scribe': 44, 'quality specialist': 31,
    'head of treasury': 12, 'warehouse officer': 43,
    'insurance officer': 33, 'application specialist': 25,
    'treasury agent': 43, 'quality controller': 31,
    'reports manager': 12,
    'operation and maintenance manager (acting)': 13,
    'after sales service manager': 12,
    'cybersecurity director': 13,
    'passenger services assistant': 51,
    'business partners': 24,
    'banking operations': 43,
}
```

```python
# Apply direct and conditional mapping rulese

def map_isco(row):
    title = row['job_title_raw']
    dept = row['department_raw']
    sector = row['sector_raw']

    if title in DIRECT_MAP:
        return DIRECT_MAP[title]

    if title == 'supervisor':
        if dept == 'sales': return 52
        if dept == 'safety & security': return 54
        if dept == 'hr': return 44
        if dept == 'technical support': return 35
        if dept in ['medical service', 'processes']: return 32
        return 33

    if title == 'sales executive':
        if sector == 'communications and it sector': return 24
        return 33

    if title == 'operations supervisor':
        if sector in ['energy sector', 'food production sector']: return 31
        if sector == 'communications and it sector': return 35
        if sector == 'medical sector': return 32
        return 33

    if title == 'trainer':
        if sector == 'education sector': return 23
        if sector == 'transport sector': return 24
        return None 

    if title == 'coordinator':
        if dept == 'hr': return 44
        if dept == 'relations': return 24
        return 33

    if title == 'consultant':
        return 24  

    if title == 'project manager':
        if sector in ['financial sector', 'media sector',
                      'engineering consulting companies']:
            return 12
        return 13  

    if title == 'lab specialist':
        if sector == 'food production sector': return 31
        if sector == 'medical sector': return 32
        return None

    return None
df['isco_2digit'] = df.apply(map_isco, axis=1)
```

```python
# Derive ISCO major group

df['isco_major'] = df['isco_2digit'].apply(
    lambda x: int(str(int(x))[0]) if pd.notna(x) else None
)
```

```python
# Map ISCO major group to major group labels

MAJOR_NAMES = {
    1: 'managers', 2: 'professionals', 3: 'technicians',
    4: 'clerical support', 5: 'services and sales',
    6: 'agricultural', 7: 'craft',
    8: 'plant and machine operators', 9: 'elementary'
}
df['major_group_isco'] = df['isco_major'].map(MAJOR_NAMES).fillna('unclassified')
```

```python
# Assign and export mapping confidence

MEDIUM_TITLES = {
    'manager assistant', 'security and safety specialist',
    'relationships specialist', 'purchase specialist',
    'relations manager', 'production supervisor', 'scribe',
    'quality specialist', 'head of treasury', 'warehouse officer',
    'insurance officer', 'application specialist', 'treasury agent',
    'quality controller', 'reports manager',
    'operation and maintenance manager (acting)',
    'after sales service manager', 'cybersecurity director',
    'passenger services assistant'
 
}

CONDITIONAL_TITLES = {
    'supervisor', 'sales executive', 'operations supervisor',
    'trainer', 'coordinator', 'consultant',
    'project manager', 'lab specialist'
}

def get_confidence(row):
    t = row['job_title_raw']
    if pd.isna(row['isco_2digit']):
        return 'unclassified'
    if t in CONDITIONAL_TITLES:
        return 'medium-conditional'
    if t in MEDIUM_TITLES:
        return 'medium'
    return 'high'

df['mapping_confidence'] = df.apply(get_confidence, axis=1)



print("Confidence:")
print(df['mapping_confidence'].value_counts())

print("\nmajor_group_isco:")
print(df['major_group_isco'].value_counts())

print(f"\nUnclassified: {df['isco_2digit'].isna().sum()} rows")
```

```python
# Check unmapped job titles

unmapped = df[df['isco_2digit'].isna()]['job_title_raw'].value_counts()
print(f"values not mapping: {len(unmapped)}")
print(unmapped)
```

```python
# Export mapping audit

mapping_table = df[[
    'job_title_raw', 'department_raw', 'sector_raw',
    'isco_2digit', 'isco_major', 'major_group_isco', 'mapping_confidence'
]].drop_duplicates()

mapping_table.to_csv('~/Desktop/Employee-Attrition- Portfolio/data/mapping_job_title.csv', index=False)
```

```python
# Logic checks
# Validate experience vs age and last-org tenure

df['logic_exp_gt_age'] = (
    df['years_experience_mid'] > (df['age_mid'] - 16)
).astype(int)

df['logic_last_gt_total'] = (
    df['years_last_org_mid'] > df['years_experience_mid']
).astype(int)

print(f"logic_exp_gt_age: {df['logic_exp_gt_age'].sum()}")
print(f"logic_last_gt_total: {df['logic_last_gt_total'].sum()}")
```

```python
# Handle missing numeric values 

exclude = ['isco_2digit', 'isco_major']
for col in df.select_dtypes(include='number').columns:
    if col in exclude:
        continue
    if df[col].isna().any():
        df[col + '_missing'] = df[col].isna().astype(int)
        df[col] = df[col].fillna(df[col].median())
print(f"Total NaN: {df.isna().sum().sum()}")
```

```python
# Drop temporary columns

drop_cols = [
    'isco_2digit_missing', 'isco_major_missing',
    'mapping_confidence',
]

df = df.drop(columns=drop_cols, errors='ignore')
```

```python
# Create ordinal training status

def training_status(row):
    if row['training_programs'] == 0:
        return 'No Training'
    elif row['useful_training'] == 0:
        return 'Trained but Not Useful'
    else:
        return 'Trained and Useful'

df['training_status'] = df.apply(training_status, axis=1)
df['training_status_ord'] = df['training_status'].map({
    'No Training': 0,
    'Trained but Not Useful': 1,
    'Trained and Useful': 2
})

print(pd.crosstab(df['training_status'], df['attrition'], margins=True))
```

```python
# Encode marital status and overtime status

df['marital_married'] = (df['marital_status_raw'] == 'married').astype(int)
df['marital_divorced'] = (df['marital_status_raw'] == 'divorced').astype(int)


# ot_1 = no overtime ot_2 = paid overtime ot_3 = unpaid overtime ot_4 = unknown
df['ot_status_label'] = df['ot_status'].map({
    1: 'no_overtime', 2: 'paid_overtime',
    3: 'unpaid_overtime', 4: 'unknown_payment'
})
df = pd.get_dummies(
    df, columns=['ot_status'],
    prefix='ot', drop_first=True, dtype=int
)

print("Marital:", [c for c in df.columns if 'marital' in c])
print("OT:", [c for c in df.columns if c.startswith('ot_')])
```
